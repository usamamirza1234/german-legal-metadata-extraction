```python


# Initialize preprocessor
preprocessor = ImagePreprocessor(r".\data\img_2.png")
output_dir = "ocr_python_textbook/temp2/"
save_intermediate = True

# Create output directory if it doesn't exist
os.makedirs(output_dir, exist_ok=True)

# 1. Inverted Images
inverted_image = preprocessor.invert_image()
if save_intermediate:
    inverted_path = f"{output_dir}inverted.jpg"
    cv2.imwrite(inverted_path, inverted_image)
    print("Text after inversion:")
    print(preprocessor.extract_fraktur_text(inverted_path))
    ##preprocessor.display(inverted_path)


```

```python
# 2. Grayscale Conversion
gray_image = preprocessor.grayscale()
if save_intermediate:
    gray_path = f"{output_dir}gray.jpg"
    cv2.imwrite(gray_path, gray_image)
    print("Text after grayscale:")
    print(preprocessor.extract_fraktur_text(gray_path))
    ##preprocessor.display(gray_path)

```

```python

# 3. Binarization
bw_image = preprocessor.binarize_image(gray_image, threshold= 127)
if save_intermediate:
    bw_path = f"{output_dir}bw_image.jpg"
    cv2.imwrite(bw_path, bw_image)
    print("Text after binarization:")
    print(preprocessor.extract_fraktur_text(bw_path))
    ##preprocessor.display(bw_path)

```

```python

# 4. Noise Removal
no_noise = preprocessor.noise_removal(bw_image)
if save_intermediate:
    no_noise_path = f"{output_dir}no_noise.jpg"
    cv2.imwrite(no_noise_path, no_noise)
    print("Text after noise removal:")
    print(preprocessor.extract_fraktur_text(no_noise_path))
    # ##preprocessor.display(no_noise_path)
```

```python
# 5. Remove Borders
no_borders = preprocessor.remove_borders(no_noise)
if save_intermediate:
    no_borders_path = f"{output_dir}no_borders.jpg"
    cv2.imwrite(no_borders_path, no_borders)
    print("Text after border removal:")
    print(preprocessor.extract_fraktur_text(no_borders_path))
    # ##preprocessor.display(no_borders_path)
```

```python

# 6. Add Borders
with_borders = preprocessor.add_borders(no_borders)
if save_intermediate:
    with_borders_path = f"{output_dir}image_with_border.jpg"
    cv2.imwrite(with_borders_path, with_borders)
    print("Text after adding borders:")
    print(preprocessor.extract_fraktur_text(with_borders_path))
    # ##preprocessor.display(with_borders_path)
```
