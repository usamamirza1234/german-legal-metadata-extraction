```python

from pdf_extractor_with_ocr import PDFExtractorWithOCR
from optimise_code.optimisation import OptimizedMetadataExtractor

debug = True
extractor = OptimizedMetadataExtractor(debug=debug)
pdf_extractor = PDFExtractorWithOCR(debug=debug)
# text = pdf_extractor.extract_text_from_pdf(pdf_path = "../pdfs/30s/30s/berufearchiv_6322.pdf", start_page=1, end_page=1)
text = pdf_extractor.extract_text_from_pdf(
    "../pdfs/30s/30s/30s_feinpolierer_1937_pruefungsanforderungen.pdf",
    start_page=1,
    preprocessing_steps=['remove_white', 'invert', 'rescale', 'binarize'],
     end_page=1,
    type="custom_preprocessing"
)


if debug:
    print("Extracted Text: ", text)

metadata = extractor.extract_all_metadata(text)
if debug:
    print("\n" + "=" * 60)
    print("EXTRACTION RESULTS:")
    print("=" * 60)

result_dict = metadata.to_dict()
print(result_dict)
for key, value in result_dict.items():
    if value and key != 'raw_text_preview':
        print(f"{key.upper()}: {value}")
```
