```python
from pdf_extractor_with_ocr import PDFExtractorWithOCR
from extractors.enhanced_extractor_metadata import EnhancedMetadataExtractor

extractor = EnhancedMetadataExtractor(debug=True)
pdf_extractor = PDFExtractorWithOCR(debug=False)


pdf_path = "pdfs/30s/30s/berufearchiv_5496.pdf"

```

```python
text = pdf_extractor.extract_text_from_pdf(pdf_path =pdf_path, start_page=1, end_page=5)
text
```

```python
metadata = extractor.extract_all_metadata(text)
```

```python
# Display results
print("\n" + "=" * 60)
print("EXTRACTED METADATA RESULTS:")
print("=" * 60)

result_dict = metadata.to_dict()
for key, value in result_dict.items():
    if value is not None and key != 'raw_text_preview':
        print(f"{key.upper()}: {value}")

print("\nCONFIDENCE SCORES:")
for field, score in metadata.confidence_scores.items():
    print(f"  {field}: {score:.2f}")
```
